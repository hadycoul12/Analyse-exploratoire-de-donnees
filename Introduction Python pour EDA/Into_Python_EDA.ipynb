En programmation, une variable est comme une boîte étiquetée qui peut contenir une valeur spécifique à un moment donné. Vous pouvez mettre quelque chose dans ce conteneur, le retirer ou le remplacer par quelque chose d'autre, et le nom de la variable fait toujours référence au contenu actuel.

Python a quelques règles et bonnes pratiques pour nommer ces variables :

* Commencer par une lettre ou un trait de soulignement
* Pas d'espace
* Sensible à la casse (les lettres minuscules sont différentes des lettres majuscules)
* Ne pas utiliser de mots réservés

Contrairement à d'autres langages de programmation, en Python, vous n'avez pas besoin d'une commande spéciale ou d'un mot-clé pour déclarer une variable. Il suffit de l'utiliser ! Vous déclarez une variable en lui donnant simplement un nom et en lui assignant une valeur à l'aide du signe égal (=).

L'instruction if indique à l'ordinateur "si une condition particulière est vraie, alors faites quelque chose". Par essence, une condition fournit une réponse Vrai ou Faux. L'instruction if utilise ensuite cette réponse, ce qui nous permet d'exécuter des actions particulières lorsque la condition est vraie.

*Imaginez que vous vous trouviez devant une porte verrouillée. Pour ouvrir cette porte, vous avez besoin de la bonne clé. Si votre clé entre dans la serrure, la porte s'ouvrira. Dans le cas contraire, la porte reste verrouillée. Ceci est très similaire à la façon dont les conditions et les instructions if fonctionnent ensemble en Python.*

L'instruction IF se compose des éléments suivants
* Le mot-clé **if** - il doit TOUJOURS être en minuscules.
* La **condition** - une expression qui s'évalue à Vrai ou Faux.
* Le **point virgule** ( : ) - Il indique la fin de la condition de l'instruction if et signifie le début du bloc de code qui sera exécuté si la condition est vraie.
* L'**indentation** - l'espace au début d'une ligne de code.

La condition peut être créée en utilisant une combinaison de variables et de comparateurs tels que >, <, >=, <=, == (égalité) ou != (différent de).

```
var_1 = 5
var_2 = 5

if var_1 == var_2:
    print("Hello !")
>>> "Hello !"
```
```
var_1 = 5
var_2 = 10

if var_1 < var_2:
    print("Hello !")
>>> "Hello !"
```

# Atterrissage

En Python, la boucle for est utilisée pour itérer sur une séquence - qui peut être une liste, un tuple ou une chaîne de caractères - ou sur d'autres objets itérables - qui peuvent être un dictionnaire, un ensemble, un fichier ou un générateur. Pendant chaque itération, un élément de la séquence est récupéré et exécuté avec le corps de la boucle.

*C'est comme une chaîne de montage qui a un nombre prédéterminé d'articles qui descendent sur le tapis roulant. Chaque travailleur (itération de la boucle) sait à l'avance combien d'articles (éléments de données) il doit traiter. Par exemple, s'il y a 100 articles sur la chaîne de montage, le travailleur effectuera sa tâche 100 fois, une fois pour chaque article.*

La boucle FOR est composée de :

*   Le **mot-clé for** - Il doit TOUJOURS être en minuscules.
*   La **variable temporaire** - Elle prend la valeur de chaque élément dans la séquence, un à la fois.
*   Le **mot-clé in** - Ce mot-clé est utilisé pour spécifier que la boucle va itérer sur la séquence donnée.
*   La **séquence** - La séquence ou l'itérable sur laquelle la boucle va itérer.
*  Et bien sûr, les **deux-points** et **l'indentation**.

Voici l'utilisation de la boucle for avec l'objet `range()`. L'objet range représente une séquence immuable d'entiers et est couramment utilisé pour boucler un nombre spécifique de fois dans les boucles for.
```
for i in range(3):
    print(i)
>>> 0
>>> 1
>>> 2
```

Ci-dessous, l'utilisation de la boucle for avec une liste.

La boucle while exécute de manière répétée un bloc de code tant qu'une condition spécifiée est vraie. Lorsque la condition devient fausse, la boucle s'arrête et le programme passe à la ligne de code suivante après la boucle. La condition est vérifiée avant chaque itération, et si elle n'est jamais vraie au départ, le bloc de code de la boucle ne s'exécutera pas du tout.

*La boucle while est comme une chaîne de montage où le travailleur ne sait pas combien d'articles vont arriver sur la ligne. Il continue simplement à travailler sur chaque article jusqu'à ce qu'une condition spécifique soit remplie, comme la fin de la journée de travail ou l'épuisement des articles sur le tapis roulant. Le travailleur (boucle) continue d'effectuer la tâche (itération) jusqu'à ce qu'une certaine condition devienne fausse, comme l'arrêt de la chaîne de montage.*

La boucle WHILE est composée de :

*   Le **mot-clé while** - Il doit TOUJOURS être en minuscules.
*   La **condition** - C'est une expression booléenne (c'est-à-dire quelque chose qui évalue soit à True soit à False). Le code à l'intérieur de la boucle s'exécute tant que cette condition est True.
*   Et bien sûr, les **deux-points** et **l'indentation**

```
a = 0
while a < 3:
    print(a)
    a = a + 1
>>> 0
>>> 1
>>> 2
```

**<font color="darkred">Étape 24. Créez une boucle WHILE qui itère tant que door_locked est True et vérifie l'état du vaisseau spatial avec une condition IF/ELIF/ELSE :**
* **<font color="darkred">si l'état de `landing_gear_deployed` est False, définissez-le sur True et affichez** *Déploiement du train d'atterrissage.*
* **si l'état de `atmosphere_suit_ready` est False, définissez-le sur True et affichez** *Préparation des combinaisons atmosphériques.*
* **si l'état de `crew_ready` est False, définissez-le sur True et affichez** *L'équipage porte maintenant ses combinaisons et est prêt.*
* **si tous les états du vaisseau spatial sont True, arrêtez la boucle WHILE et affichez** *Toutes les conditions remplies. Ouverture de la porte vers Mars.*

*Message de l'Équipage : Après l'atterrissage, la tâche immédiate est de se préparer à la sortie.<br> Cela implique le déploiement du train d'atterrissage, l'enfilage des combinaisons et la vérification de la disponibilité de l'équipage.<br> Nous devons vérifier que tout est prêt avant d'ouvrir la porte..*

````python
spacecraft_status = {
    'door_locked': True,
    'landing_gear_deployed': False,
    'atmosphere_suit_ready': False,
    'crew_ready': False
}
````

````python
while spacecraft_status["door_locked"]:
    if not spacecraft_status["landing_gear_deployed"] :
        spacecraft_status["landing_gear_deployed"] = True
        print("Deployment of the landing gear.")
    elif not spacecraft_status["atmosphere_suit_ready"] :
        spacecraft_status["atmosphere_suit_ready"] = True
        print("Preparing the atmosphere suits.")
    elif not spacecraft_status["crew_ready"] :
        spacecraft_status["crew_ready"] = True
        print("The crew is now wearing their suits and is ready.")
    else :
        spacecraft_status["door_locked"] = False
        print("All conditions met. Opening door to Mars.")
````
